## 4. Duplicate records

```python
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate emails (potential candidates who applied more than once):", df['Email'].duplicated().sum())
```

## 5. Unique values in relevant categorical attributes

```python
for col in ['Country', 'Seniority', 'Technology']:
    print(f"{col}: {df[col].nunique()} unique values")

print()
print("Seniority Levels:")
print(sorted(df['Seniority'].unique()))
```

## 6. Application date range

```python
df['Application Date'] = pd.to_datetime(df['Application Date'])
print("Min date:", df['Application Date'].min().date())
print("Max date:", df['Application Date'].max().date())
```

## 7. Score range and descriptive statistics

```python
df[['YOE', 'Code Challenge Score', 'Technical Interview Score']].describe()
```

## 8. Business rule preview
For exploratory purposes only.

```python
hired = (df['Code Challenge Score'] >= 7) & (df['Technical Interview Score'] >= 7)
print("Hired candidates:", hired.sum())
print(f"Hiring rate: {hired.mean()*100:.2f}%")
```

## Main findings

* The dataset has **50,000 rows and 10 columns**, with no null values in any column.
* The file uses `;` as a separator (not `,`) this needs to be kept in mind during the Extract step.
* There are no fully duplicate rows, but **167 emails are repeated** → some candidates applied more than once. This confirms that the grain of the dimensional model must be "one application", not "a unique candidate".
* There are **244 distinct countries** and **24 distinct technologies**.
* `Seniority` has **7 levels**: Architect, Intern, Junior, Lead, Mid-Level, Senior, Trainee.
* `YOE` ranges from **0 to 30 years** (average ~15.3), and is not correlated with Seniority (they are independent attributes).
* Both scores (`Code Challenge Score` and `Technical Interview Score`) range from **0 to 10**.
* Application dates range from **2018-01-01 to 2022-07-04**.
* Under the HIRED business rule, the overall hiring rate is **~13.4%**.
